```python
def find_two_sum(nums, target):
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return (i, j), (nums[i], nums[j])  # Return indices and numbers
    return "No valid pair found"

# User input
nums = list(map(int, input("Enter numbers separated by spaces: ").split()))
target = int(input("Enter target sum: "))

# Find and print result
result = find_two_sum(nums, target)
print("Output:", result)
```

```python
from collections import Counter

def word_frequency(words):
    freq = Counter(words)  # Count occurrences of each word
    sorted_words = sorted(freq.items(), key=lambda x: x[1], reverse=True)  # Sort by frequency (descending)
    
    unique_words = [word for word, count in sorted_words]  # Extract words only
    return unique_words, sorted_words  # Return unique words and frequencies

# User input
words = input("Enter words separated by spaces: ").split()

# Get unique words and their frequencies
unique_words, sorted_frequencies = word_frequency(words)

# Print results
print("Unique words in descending frequency order:", unique_words)
print("Word frequencies:")
for word, count in sorted_frequencies:
    print(f"{word}: {count}")
```

```python
def longest_consecutive_sequence(nums):
    if not nums:
        return []  # Return empty list if input is empty

    nums.sort()  # Sort the list
    longest = []
    current = [nums[0]]  # Start with the first number

    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1] + 1:  # Check if consecutive
            current.append(nums[i])
        elif nums[i] != nums[i - 1]:  # Ignore duplicates
            if len(current) > len(longest):  # Update longest sequence
                longest = current
            current = [nums[i]]  # Start a new sequence

    if len(current) > len(longest):  # Final check
        longest = current

    return longest

# User input
nums = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Find and print the longest consecutive sequence
result = longest_consecutive_sequence(nums)
print("Longest consecutive sequence:", result)
```

```python
# Open and read the file
def count_hours(filename):
    hour_counts = {}  # Dictionary to store hour frequencies

    with open(filename, "r") as file:
        for line in file:
            if line.startswith("From "):  # Only process lines starting with "From "
                words = line.split()  # Split line into words
                time = words[5]  # Extract the timestamp (6th word)
                hour = time.split(":")[0]  # Get the hour (first part of time)
                hour_counts[hour] = hour_counts.get(hour, 0) + 1  # Count occurrences

    # Sort dictionary by hour and print results
    for hour, count in sorted(hour_counts.items()):
        print(hour, count)

# Run the function with "Data.txt"
count_hours("Data.txt")
```

```python
# Price dictionary for menu items
menu_prices = {
    "frozen_yogurt": {
        "s": {"v": 3.09, "u": 3.19},
        "m": {"v": 3.89, "u": 3.99},
        "l": {"v": 4.69, "u": 4.79}
    },
    "ice_cream": {
        "s": {"v": 3.49, "u": 3.59},
        "m": {"v": 4.49, "u": 4.59},
        "l": {"v": 5.49, "u": 5.59}
    },
    "toppings": {"g": 1.59, "n": 1.89},
    "cookies": {
        "r": {"r": 1.49, "l": 2.49},
        "c": {"r": 1.59, "l": 2.59}
    }
}

# Dictionary to track customer spending
customer_totals = {}

# Dictionary to track order stats
order_stats = {"frozen_yogurt": {"count": 0, "total": 0}, "ice_cream": {"count": 0, "total": 0}}

# Main ordering loop
while True:
    name = input("\nEnter your name: ").strip().title()
    
    # Check if customer exists
    if name in customer_totals:
        print("Welcome back!")
    else:
        customer_totals[name] = 0

    # Order type (Frozen Yogurt or Ice Cream)
    order_type = input("You want Frozen Yogurt or Ice Cream? (Y: Frozen Yogurt, I: Ice Cream) ").strip().lower()
    if order_type == "y":
        category = "frozen_yogurt"
    elif order_type == "i":
        category = "ice_cream"
    else:
        print("Invalid choice, try again.")
        continue

    # Size (Small, Medium, Large)
    size = input("What size? (S: Small, M: Medium, L: Large) ").strip().lower()
    if size not in menu_prices[category]:
        print("Invalid size, try again.")
        continue

    # Container (Cone or Cup)
    container = input("What kind of container? (V: Cone, U: Cup) ").strip().lower()
    if container not in menu_prices[category][size]:
        print("Invalid container, try again.")
        continue

    order_total = menu_prices[category][size][container]  # Base price

    # Toppings
    add_topping = input("Do you like to add a topping? (Y: Yes, N: No) ").strip().lower()
    if add_topping == "y":
        topping = input("What topping do you like? (G: Gummies, N: Nuts) ").strip().lower()
        if topping in menu_prices["toppings"]:
            order_total += menu_prices["toppings"][topping]
        else:
            print("Invalid topping, skipping.")

    # Cookies
    add_cookies = input("Do you like to have cookies? (Y: Yes, N: No) ").strip().lower()
    if add_cookies == "y":
        cookie_type = input("What kind of cookies? (R: Raisin, C: Chocolate) ").strip().lower()
        cookie_size = input("And the size? (R: Regular, L: Large) ").strip().lower()
        if cookie_type in menu_prices["cookies"] and cookie_size in menu_prices["cookies"][cookie_type]:
            order_total += menu_prices["cookies"][cookie_type][cookie_size]
        else:
            print("Invalid cookie choice, skipping.")

    # Get day of the week
    day = int(input("What day of the week is this? (1: Mon, 2: Tue, ..., 7: Sun) ").strip())
    if day in [6, 7]:  # Apply weekend discount (10% off)
        order_total *= 0.9

    # Update customer total
    customer_totals[name] += round(order_total, 2)

    # Update order stats
    order_stats[category]["count"] += 1
    order_stats[category]["total"] += order_total

    print(f"\nYour order total is ${round(order_total, 2)}")

    # Continue or Quit
    choice = input("Enter C to continue or Q to quit: ").strip().lower()
    if choice == "q":
        break

# Print Sales Report
print("\n*************** Sales Report ***************")
print("Customer Name     Order Total")
for customer, total in customer_totals.items():
    print(f"{customer:<15} ${total:.2f}")

print("\nItem            # Orders   Order Total   Average per Order")
for item, data in order_stats.items():
    avg_order = data["total"] / data["count"] if data["count"] > 0 else 0
    print(f"{item.replace('_', ' ').title():<15} {data['count']:<10} ${data['total']:.2f}      ${avg_order:.2f}")

grand_total = sum(data["total"] for data in order_stats.values())
grand_orders = sum(data["count"] for data in order_stats.values())
avg_grand = grand_total / grand_orders if grand_orders > 0 else 0

print(f"\nGrand Total      {grand_orders:<10} ${grand_total:.2f}      ${avg_grand:.2f}")
```

```python
import random

QUESTIONS_FILE = "quest_load.txt"
LEADERBOARD_FILE = "leaders.txt"

# Load questions from file
def load_questions():
    questions = {}
    try:
        with open(QUESTIONS_FILE, "r") as file:
            for line in file:
                parts = line.strip().split("|")
                if len(parts) == 3:
                    question, answer, points = parts
                    questions[question] = (answer.lower(), int(points))
        print("Questions loaded successfully!")
    except FileNotFoundError:
        print(f"Error: {QUESTIONS_FILE} not found.")
    return questions

# Add a question to the file
def add_question():
    question = input("Enter the question: ").strip()
    answer = input("Enter the answer: ").strip()
    points = input("Enter the points for this question: ").strip()
    with open(QUESTIONS_FILE, "a") as file:
        file.write(f"{question}|{answer}|{points}\n")
    print("Question added successfully!")

# Show all loaded questions
def show_questions(questions):
    if not questions:
        print("No questions loaded. Please load questions first.")
        return
    print("\n--- Available Questions ---")
    for q, (a, p) in questions.items():
        print(f"{q} ({p} points)")

# Load leaderboard from file
def load_leaderboard():
    leaderboard = {}
    try:
        with open(LEADERBOARD_FILE, "r") as file:
            for line in file:
                name, score = line.strip().split("|")
                leaderboard[name] = int(score)
    except FileNotFoundError:
        print(f"Warning: {LEADERBOARD_FILE} not found. Creating new leaderboard.")
    return leaderboard

# Save leaderboard to file
def save_leaderboard(leaderboard):
    with open(LEADERBOARD_FILE, "w") as file:
        for name, score in leaderboard.items():
            file.write(f"{name}|{score}\n")

# Play the trivia game
def play_game(questions, leaderboard):
    if not questions:
        print("No questions loaded. Please load questions first.")
        return
    
    name = input("Enter your first name: ").strip()
    
    if name in leaderboard:
        print(f"Welcome back, {name}! Your previous score was {leaderboard[name]}")
    else:
        print(f"Welcome to Trivia Madness, {name}!")
        leaderboard[name] = 0

    num_questions = int(input(f"How many questions do you want to try? (Max {len(questions)}): "))
    num_questions = min(num_questions, len(questions))

    selected_questions = random.sample(list(questions.keys()), num_questions)
    score = 0

    for q in selected_questions:
        answer = input(f"{q}\nYour answer: ").strip().lower()
        correct_answer, points = questions[q]
        if answer == correct_answer:
            print("Correct!")
            score += points
        else:
            print(f"Incorrect! The correct answer was: {correct_answer}")

    leaderboard[name] += score
    print(f"Your total score is now: {leaderboard[name]}")
    save_leaderboard(leaderboard)

# Show the leaderboard
def show_score():
    leaderboard = load_leaderboard()
    sorted_leaderboard = sorted(leaderboard.items(), key=lambda x: x[1], reverse=True)

    print("\n<<< LEADER BOARD >>>")
    for name, score in sorted_leaderboard:
        print(f"{name} {score}")

# Main menu
def main():
    questions = {}
    leaderboard = load_leaderboard()

    while True:
        print("\nPlease select an item from the menu below:")
        print("1- Load Questions")
        print("2- Add Question")
        print("3- Show Questions")
        print("4- Play Game")
        print("5- Show Score")
        print("6- Quit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            questions = load_questions()
        elif choice == "2":
            add_question()
        elif choice == "3":
            show_questions(questions)
        elif choice == "4":
            play_game(questions, leaderboard)
        elif choice == "5":
            show_score()
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 6.")

if __name__ == "__main__":
    main()
```
